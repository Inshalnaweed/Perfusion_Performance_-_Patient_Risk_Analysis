```python
import pandas as pd
import matplotlib.pyplot as plt

# Upload your file
from google.colab import files
uploaded = files.upload()

# Read CSV file
df = pd.read_csv(list(uploaded.keys())[0])


# 1. CPB Time vs ICU Stay
plt.figure(figsize=(8,5))
plt.scatter(df["CPB_Time"], df["ICU_Stay"])
plt.xlabel("CPB Time (minutes)")
plt.ylabel("ICU Stay (days)")
plt.title("CPB Time vs ICU Stay")
plt.show()

# 2. Surgery Type Distribution
plt.figure(figsize=(8,5))
df["Surgery_Type"].value_counts().plot(kind="bar")
plt.xlabel("Surgery Type")
plt.ylabel("Number of Patients")
plt.title("Number of Patients by Surgery Type")
plt.xticks(rotation=45)
plt.show()


# 3. ICU Stay Distribution
plt.figure(figsize=(8,5))
plt.hist(df["ICU_Stay"], bins=10)
plt.xlabel("ICU Stay (days)")
plt.ylabel("Number of Patients")
plt.title("Distribution of ICU Stay")
plt.show()


# 4. CPB Time vs Complication
plt.figure(figsize=(8,5))
df.boxplot(column="CPB_Time", by="Complication")
plt.xlabel("Complication")
plt.ylabel("CPB Time (minutes)")
plt.title("CPB Time by Complication Status")
plt.suptitle("")
plt.show()


# 5. Lactate vs ICU Stay
plt.figure(figsize=(8,5))
plt.scatter(df["Lactate_Peak"], df["ICU_Stay"])
plt.xlabel("Peak Lactate")
plt.ylabel("ICU Stay (days)")
plt.title("Peak Lactate vs ICU Stay")
plt.show()
```
